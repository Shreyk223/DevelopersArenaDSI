## 🧹 Phase 2: Data Collection & Preparation
We will load the separate datasets, fix their primary key formatting (`C00001` vs `CUST001`), execute a `LEFT JOIN` to maintain our 500-row base, and handle missing values.

```python
# Load Data
try:
    churn = pd.read_csv('data/customer_churn.csv')
    sales = pd.read_csv('data/sales_data.csv')
except:
    churn = pd.read_csv('customer_churn.csv')
    sales = pd.read_csv('sales_data.csv')

# Create matching Primary Keys for merging
churn['CustNum'] = churn['CustomerID'].str.replace('C', '').astype(int)
sales['CustNum'] = sales['Customer_ID'].str.replace('CUST', '').astype(int)

# LEFT JOIN: Keep all 500 churn records, attach sales data where available
df = pd.merge(churn, sales, on='CustNum', how='left')

# Clean Data: Impute NaNs for customers who didn't make a hardware purchase
df['Total_Sales'].fillna(0, inplace=True)
df['Quantity'].fillna(0, inplace=True)
df['Product'].fillna('Subscription Only', inplace=True)
df['Region'].fillna('Unknown', inplace=True)

print(f"Merged Dataset Shape: {df.shape} -> Requirement of 500+ rows met!")
import os
os.makedirs('data', exist_ok=True)
df.to_csv('data/cleaned_merged_data.csv', index=False)
print("✅ Data Cleaned and Merged Successfully.")
df.head()
```
